```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import cv2
import tensorflow as tf
from tqdm import tqdm
import os
from sklearn.utils import shuffle
from sklearn.model_selection import train_test_split
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras.callbacks import ReduceLROnPlateau, TensorBoard, ModelCheckpoint
from sklearn.metrics import classification_report,confusion_matrix
from PIL import Image
```

---

# Color

```python
colors_dark = ["#1F1F1F", "#313131", '#636363', '#AEAEAE', '#DADADA']
colors_red = ["#331313", "#582626", '#9E1717', '#D35151', '#E9B4B4']
colors_green = ['#01411C','#4B6F44','#4F7942','#74C365','#D0F0C0']

sns.palplot(colors_dark)
sns.palplot(colors_green)
sns.palplot(colors_red)
```

---

# Data Preperation

```python
labels = ['Fake','Real']
```

```python
DATASET_PATH = r"D:\DeepFake Detection\ train\ data"
```

```python
X_train = []    #actual images
y_train = []    #labels F/R
image_size = 150
for i in labels:
    folderPath = os.path.join(DATASET_PATH,'Train',i)
    for j in tqdm(os.listdir(folderPath)):
        img = cv2.imread(os.path.join(folderPath,j))
        img = cv2.resize(img,(image_size, image_size))
        X_train.append(img)
        y_train.append(i)

for i in labels:
    folderPath = os.path.join(DATASET_PATH,'Test',i)
    for j in tqdm(os.listdir(folderPath)):
        img = cv2.imread(os.path.join(folderPath,j))
        img = cv2.resize(img,(image_size,image_size))
        X_train.append(img)
        y_train.append(i)

X_train = np.array(X_train)
y_train = np.array(y_train)
```

```python
X_train, y_train = shuffle(X_train,y_train, random_state=42)
```

```python
X_train.shape
```

Dividing the dataset into **Training** and **Testing** sets.

```python
X_train,X_test,y_train,y_test = train_test_split(X_train,y_train, test_size=0.1,random_state=42)
```

```python
y_train_new = []
for i in y_train:
    y_train_new.append(labels.index(i))
y_train = y_train_new
y_train = tf.keras.utils.to_categorical(y_train)


y_test_new = []
for i in y_test:
    y_test_new.append(labels.index(i))
y_test = y_test_new
y_test = tf.keras.utils.to_categorical(y_test)
```

---

# Transfer Learning

```python
effnet = EfficientNetB0(weights='imagenet',include_top=False,input_shape=(image_size,image_size,3))
```

weights='imagenet' - Load the weights that EfficientNetB0 learned from the ImageNet dataset.
include_top=False - The original EfficientNetB0 was designed to classify ImageNet images into 1000 classes. we need 2 classes- fake real.

EfficientNetB0 as the base CNN and adding my own classification layers on top.

```python
model = effnet.output
model = tf.keras.layers.GlobalAveragePooling2D()(model)
model = tf.keras.layers.Dropout(rate=0.5)(model)
model = tf.keras.layers.Dense(2,activation='softmax')(model)
model = tf.keras.models.Model(inputs=effnet.input, outputs = model)
```

```python
model.summary()
```

```python
model.compile(loss='categorical_crossentropy',optimizer = 'Adam', metrics= ['accuracy'])
#Adam is the algorithm that updates the neural network's weights during training.
```

Callbacks are Instructions that TensorFlow follows automatically while the model is training. 
TensorBoard records training information
ModelCheckpoint automatically saves your model when it finds a better validation accuracy
Epoch 3 → 94.6%  (save)
ReduceLROnPlateau - If validation accuracy stops improving, reduce the learning rate.

```python
tensorboard = TensorBoard(log_dir = 'logs')
checkpoint = ModelCheckpoint("effnet.keras",monitor="val_accuracy",save_best_only=True,mode="auto",verbose=1)
reduce_lr = ReduceLROnPlateau(monitor = 'val_accuracy', factor = 0.3, patience = 2, min_delta = 0.001,
                              mode='auto',verbose=1)
```

---

# Training The Model

```python
history = model.fit(X_train,y_train,validation_split=0.1, epochs =5, verbose=1, batch_size=32)
```

```python
model.save('my_model.keras')
```

```python
from warnings import filterwarnings
```

```python
filterwarnings('ignore')

epochs = [i for i in range(5)]  #one complete pass through the training dataset.
fig, ax = plt.subplots(1,2,figsize=(14,7))  #1row, 2cols
train_acc = history.history['accuracy']  #graph1
train_loss = history.history['loss']  #graph2
val_acc = history.history['val_accuracy']   #history is the object returned when you trained your TensorFlow/Keras model,
val_loss = history.history['val_loss']

fig.text(s='Epochs vs. Training and Validation Accuracy/Loss',size=18,fontweight='bold',
             fontname='monospace',color=colors_dark[1],y=1,x=0.28,alpha=0.8)

sns.despine()  #Remove unnecessary borders

#graph1
ax[0].plot(epochs, train_acc, marker='o',  markerfacecolor=colors_green[2],color=colors_green[3],
           label = 'Training Accuracy')
ax[0].plot(epochs, val_acc, marker='o',markerfacecolor=colors_red[2],color=colors_red[3],
           label = 'Validation Accuracy')
ax[0].legend(frameon=False)
ax[0].set_xlabel('Epochs')
ax[0].set_ylabel('Accuracy')

sns.despine()

#graph2
ax[1].plot(epochs, train_loss, marker='o',markerfacecolor=colors_green[2],color=colors_green[3],
           label ='Training Loss')
ax[1].plot(epochs, val_loss, marker='o',markerfacecolor=colors_red[2],color=colors_red[3],
           label = 'Validation Loss')
ax[1].legend(frameon=False)
ax[1].set_xlabel('Epochs')
ax[1].set_ylabel('Training & Validation Loss')

fig.show()

```

---

# Prediction

```python
pred = model.predict(X_test)
pred = np.argmax(pred,axis=1)   #takes class with high prob. F/R
y_test_new = np.argmax(y_test,axis=1)
```

---

# Evaluation

```python
print(classification_report(y_test_new,pred))
```

```python
fig,ax=plt.subplots(1,1,figsize=(14,7))

sns.heatmap(confusion_matrix(y_test_new,pred),ax=ax,xticklabels=labels,yticklabels=labels,annot=True,
           cmap=colors_green[::-1],alpha=0.7,linewidths=2,linecolor=colors_dark[3])
fig.text(s='Heatmap of the Confusion Matrix',size=18,fontweight='bold',
             fontname='monospace',color=colors_dark[1],y=0.92,x=0.28,alpha=0.8)

plt.show()   
```
